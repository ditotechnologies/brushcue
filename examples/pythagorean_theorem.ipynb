```python
import brushcue as bc

a = 3.0
b = 4.0
graph = bc.Float.pow((bc.Float.pow(a, 2.0) + bc.Float.pow(b, 2.0)), 0.5)

ctx = bc.Context()
print(graph.execute(ctx))
```
