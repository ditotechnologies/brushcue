```python
import brushcue
from PIL import Image
import io

input_image_1 = brushcue.composition_monet_women_with_parasol() # insert your own image here
float_multiply_2 = brushcue.float_multiply(2000.0, -0.000000022351742)
float_add_3 = brushcue.float_add(float_multiply_2, 5003.0)
warmth_4 = brushcue.composition_target_white_kelvin(input_image_1, float_add_3)

ctx = brushcue.Context()
result = warmth_4.execute(ctx)
composition = result.as_composition()
data_bytes = composition.to_image_bytes(ctx)
img = Image.open(io.BytesIO(data_bytes))
img.thumbnail((400, 400)) # remove this line for full resolution
img
```
