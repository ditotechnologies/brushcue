```python
import brushcue
from PIL import Image
import io

input_image_1 = brushcue.composition_monet_women_with_parasol() # insert your own image here
float_constant_2 = brushcue.float_constant(5003.0)
float_constant_3 = brushcue.float_constant(2000.0)
float_constant_4 = brushcue.float_constant(-2.2351741790771484e-8)
factor_5 = brushcue.float_passthrough(float_constant_4)
float_multiply_6 = brushcue.float_multiply(float_constant_3, factor_5)
float_add_7 = brushcue.float_add(float_multiply_6, float_constant_2)
warmth_8 = brushcue.composition_target_white_kelvin(input_image_1, float_add_7)
result_image_9 = brushcue.composition_passthrough(warmth_8)

ctx = brushcue.Context()
result = result_image_9.execute(ctx)
composition = result.as_composition()
data_bytes = composition.to_image_bytes(ctx)
img = Image.open(io.BytesIO(data_bytes))
img
```
