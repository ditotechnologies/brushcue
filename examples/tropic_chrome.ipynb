```python
import io
from PIL import Image

import brushcue as bc

strength = 1.0
input_image = bc.Composition.monet_women_with_parasol()
bounds = input_image.bounds()
graded = (
    input_image.r_g_b_curve(
        bc.Curve.gamma(0.97), bc.Curve.gamma(0.95), bc.Curve.gamma(0.97)
    )
    .l_curve(bc.Curve.s_curve(0.5, 1.2, 1.0, 1.0))
    .saturation_adjust(1.3)
)
filtered = graded.film_grain(1.51, 150.0, 0.5, 51.0, 0.3, 300.0, 0.2).crop(bounds)
graph = filtered.opacity_scales(strength).blend_alpha(input_image).crop(bounds)

ctx = bc.Context()
composition = graph.execute(ctx)
data_bytes = composition.to_image_bytes(ctx)
img = Image.open(io.BytesIO(data_bytes))
img.thumbnail((400, 400))  # Remove this line for full resolution
img
```
