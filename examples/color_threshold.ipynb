```python
import brushcue
from PIL import Image
import io

input_image_1 = brushcue.composition_monet_women_with_parasol() # insert your own image here
float_constant_2 = brushcue.float_constant(0.14500001072883606)
threshold_3 = brushcue.float_passthrough(float_constant_2)
color_threshold_4 = brushcue.composition_color_threshold(input_image_1, threshold_3)

ctx = brushcue.Context()
result = color_threshold_4.execute(ctx)
composition = result.as_composition()
data_bytes = composition.to_image_bytes(ctx)
img = Image.open(io.BytesIO(data_bytes))
img.thumbnail((400, 400)) # remove this line for full resolution
img
```
