```python
import brushcue as bc

duration = 3.0


@bc.brushcue_fn
def frame(time):
    return bc.Composition.monet_women_with_parasol().kaleidoscope(
        4.0, ((time / duration) * bc.Float.pi()), 0.09, 0.0
    )


graph = bc.Sequence.graph(duration, frame)

ctx = bc.Context()
sequence = graph.execute(ctx)
# Render `sequence` to a video file as needed.
```
