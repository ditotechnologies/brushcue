```python
import brushcue
from PIL import Image
import io

input_image_1 = brushcue.composition_monet_women_with_parasol() # insert your own image here
composition_bounds_2 = brushcue.composition_bounds(input_image_1)
composition_bounds_3 = brushcue.composition_bounds(input_image_1)
composition_bounds_4 = brushcue.composition_bounds(input_image_1)
composition_bounds_5 = brushcue.composition_bounds(input_image_1)
composition_bounds_6 = brushcue.composition_bounds(input_image_1)
bounds2f_width_7 = brushcue.bounds2f_width(composition_bounds_3)
bounds2f_height_8 = brushcue.bounds2f_height(composition_bounds_4)
bounds2f_width_9 = brushcue.bounds2f_width(composition_bounds_5)
bounds2f_height_10 = brushcue.bounds2f_height(composition_bounds_6)
float_divide_11 = brushcue.float_divide(bounds2f_width_7, 2.0)
float_divide_12 = brushcue.float_divide(bounds2f_height_8, 2.0)
float_divide_13 = brushcue.float_divide(bounds2f_width_9, 2.0)
float_divide_14 = brushcue.float_divide(bounds2f_height_10, 2.0)
vector_2_float_from_components_15 = brushcue.vector2f_from_components(float_divide_13, float_divide_14)
composition_vignette_16 = brushcue.composition_vignette(input_image_1, vector_2_float_from_components_15, float_divide_11, float_divide_12, 500.0, 0.8)
composition_crop_17 = brushcue.composition_crop(composition_vignette_16, composition_bounds_2)

ctx = brushcue.Context()
result = composition_crop_17.execute(ctx)
composition = result.as_composition()
data_bytes = composition.to_image_bytes(ctx)
img = Image.open(io.BytesIO(data_bytes))
img.thumbnail((400, 400)) # remove this line for full resolution
img
```
