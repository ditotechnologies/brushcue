```python
import brushcue
from PIL import Image
import io

input_image_1 = brushcue.composition_monet_women_with_parasol() # insert your own image here
float_constant_2 = brushcue.float_constant(0.25)
float_constant_3 = brushcue.float_constant(0.30000001192092896)
float_constant_4 = brushcue.float_constant(0.949999988079071)
radius_5 = brushcue.float_passthrough(float_constant_2)
softness_6 = brushcue.float_passthrough(float_constant_3)
strength_7 = brushcue.float_passthrough(float_constant_4)
composition_vignette_8 = brushcue.composition_vignette(input_image_1, radius_5, softness_6, strength_7)

ctx = brushcue.Context()
result = composition_vignette_8.execute(ctx)
composition = result.as_composition()
data_bytes = composition.to_image_bytes(ctx)
img = Image.open(io.BytesIO(data_bytes))
img.thumbnail((400, 400)) # remove this line for full resolution
img
```
