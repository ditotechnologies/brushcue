```python
import brushcue
from PIL import Image
import io

input_image_1 = brushcue.composition_monet_women_with_parasol() # insert your own image here (1739x2160)
center_2 = brushcue.vector2f_from_components(870.0, 1080.0)
composition_vignette_3 = brushcue.composition_vignette(input_image_1, center_2, 550.0, 700.0, 400.0, 0.95)

ctx = brushcue.Context()
result = composition_vignette_3.execute(ctx)
composition = result.as_composition()
data_bytes = composition.to_image_bytes(ctx)
img = Image.open(io.BytesIO(data_bytes))
img.thumbnail((400, 400)) # remove this line for full resolution
img
```
