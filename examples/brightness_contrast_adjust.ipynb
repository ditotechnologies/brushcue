```python
import brushcue
from PIL import Image
import io

input_image_1 = brushcue.composition_monet_women_with_parasol() # insert your own image here
float_constant_2 = brushcue.float_constant(0.5)
float_constant_3 = brushcue.float_constant(2.0)
brightness_4 = brushcue.float_passthrough(float_constant_3)
curve_pivoted_sigmoid_5 = brushcue.curve_pivoted_sigmoid(float_constant_2, brightness_4)
brightness_contrast_adjustment_6 = brushcue.composition_l_curve(input_image_1, curve_pivoted_sigmoid_5)

ctx = brushcue.Context()
result = brightness_contrast_adjustment_6.execute(ctx)
composition = result.as_composition()
data_bytes = composition.to_image_bytes(ctx)
img = Image.open(io.BytesIO(data_bytes))
img.thumbnail((400, 400)) # remove this line for full resolution
img
```
