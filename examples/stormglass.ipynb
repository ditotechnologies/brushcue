```python
import io
from PIL import Image

import brushcue as bc


@bc.brushcue_fn
def apply(input_image, strength):
    graded = input_image.color_transformer_shader(
        "let l = clamp(input.x, 0.0, 1.0);\nlet shadow = 1.0 - smoothstep(0.16, 0.62, l);\nlet red = smoothstep(0.035, 0.12, input.y);\nlet light = smoothstep(0.63, 0.98, l);\nlet tone = l + 0.12 * (l - 0.45) * sin(l * 3.14159265);\nlet a = input.y * (0.80 + 0.48 * red) - 0.026 * shadow;\nlet b = input.z * (1.0 - 0.35 * light) - 0.037 * shadow - 0.024 * red;\nreturn vec4f(mix(input.xyz, vec3f(tone, a, b), amount), input.w);",
        "",
        bc.ColorRepresentation.oklab_a(),
        bc.ColorRepresentation.oklab_a(),
        bc.Dictionary.create().add("amount", strength),
    )
    sampling = graded.color_transformer_shader(
        "return vec4f(input.rgb * input.a, input.a);",
        "",
        bc.ColorRepresentation.srgb(),
        bc.ColorRepresentation.srgb(),
        bc.Dictionary.create(),
    )
    bounds = input_image.bounds()
    reference_scale = (bounds.width().min(bounds.height()) / 1000.0).max(0.001)
    radius = (2.2 * reference_scale).max(0.5)
    return sampling.spacial_effect_shader(
        "let lo = image_bounds.xy + vec2f(0.5);\nlet hi = image_bounds.zw - vec2f(0.5);\nlet center_sample = sample(clamp(position, lo, hi));\nlet center = vec4f(center_sample.rgb / max(center_sample.a, 0.00001), center_sample.a);\nif (center.a <= 0.0) { return center; }\nvar total = vec3f(0.0);\nvar weight = 0.0;\nfor (var y = -1; y <= 1; y = y + 1) {\n  for (var x = -1; x <= 1; x = x + 1) {\n    let offset = vec2f(f32(x), f32(y));\n    let raw = sample(clamp(position + offset * radius, lo, hi));\n    let c = vec4f(raw.rgb / max(raw.a, 0.00001), raw.a);\n    let w = c.a;\n    total += c.rgb * w; weight += w;\n  }\n}\nlet local = total / max(weight, 0.00001);\nlet edge = clamp(center.rgb - local, vec3f(0.0), vec3f(0.12));\nlet light = smoothstep(0.40, 0.85, max(center.r, max(center.g, center.b)));\nreturn vec4f(center.rgb + edge * light * 0.75 * amount, center.a);",
        "",
        radius,
        bc.Dictionary.create()
        .add("amount", strength)
        .add("radius", radius)
        .add("image_bounds", bounds)
        .add("reference_scale", reference_scale),
        bc.ColorRepresentation.srgb(),
    ).crop(bounds)


strength = 1.0
input_image = bc.Composition.monet_women_with_parasol()
graph = apply(input_image, strength)

ctx = bc.Context()
composition = graph.execute(ctx)
data_bytes = composition.to_image_bytes(ctx)
img = Image.open(io.BytesIO(data_bytes))
img.thumbnail((400, 400))  # Remove this line for full resolution
img
```
