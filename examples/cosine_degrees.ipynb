```python
import brushcue as bc

angle_degrees = 60.0
graph = bc.Float.cos(
    bc.Float.divide(bc.Float.multiply(angle_degrees, bc.Float.pi()), 180.0)
)

ctx = bc.Context()
print(graph.execute(ctx))
```
