```python
import brushcue as bc

semi_major_axis = 4.0
semi_minor_axis = 3.0
graph = bc.Float.multiply(
    bc.Float.pi(), bc.Float.multiply(semi_major_axis, semi_minor_axis)
)

ctx = bc.Context()
print(graph.execute(ctx))
```
