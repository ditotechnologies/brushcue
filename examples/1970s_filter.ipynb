```python
import io
from PIL import Image

import brushcue as bc

input_image = bc.Composition.monet_women_with_parasol()
graded = (
    input_image.r_g_b_curve(
        bc.Curve.gamma(0.9), bc.Curve.gamma(0.97), bc.Curve.gamma(1.14)
    )
    .l_curve(bc.Curve.pivoted_sigmoid(0.52, 0.86))
    .saturation_adjust(0.86)
)
bounds = input_image.bounds()
filtered = graded.film_grain(1.1, 150.0, 0.5, 51.0, 0.3, 300.0, 0.16).crop(bounds)
strength = 1.0
graph = (
    filtered.opacity_scales(strength)
    .blend_alpha(input_image, bc.Transform2.identity())
    .crop(bounds)
)

ctx = bc.Context()
composition = graph.execute(ctx)
data_bytes = composition.to_image_bytes(ctx)
img = Image.open(io.BytesIO(data_bytes))
img.thumbnail((400, 400))  # Remove this line for full resolution
img
```
