```python
import brushcue
from PIL import Image
import io

input_image_1 = brushcue.composition_monet_women_with_parasol() # insert your own image here
vector_2_float_from_components_2 = brushcue.vector2f_from_components(-0.000000018626451, -0.000000018626451)
composition_chroma_offset_3 = brushcue.composition_chroma_offset(input_image_1, vector_2_float_from_components_2)

ctx = brushcue.Context()
result = composition_chroma_offset_3.execute(ctx)
composition = result.as_composition()
data_bytes = composition.to_image_bytes(ctx)
img = Image.open(io.BytesIO(data_bytes))
img.thumbnail((400, 400)) # remove this line for full resolution
img
```
