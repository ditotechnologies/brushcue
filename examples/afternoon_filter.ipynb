```python
import io
from PIL import Image

import brushcue as bc

input_image = bc.Composition.monet_women_with_parasol()
intensity = 0.6
graph = (
    input_image.target_white_kelvin((5003.0 - (700.0 * intensity)))
    .chroma_offset(bc.Vector2f.from_components((0.005 * intensity), (0.02 * intensity)))
    .contrast_adjustment((0.1 * intensity))
    .saturation_adjust((1.0 + (0.1 * intensity)))
    .crop(input_image.bounds())
)

ctx = bc.Context()
composition = graph.execute(ctx)
data_bytes = composition.to_image_bytes(ctx)
img = Image.open(io.BytesIO(data_bytes))
img.thumbnail((400, 400))  # Remove this line for full resolution
img
```
