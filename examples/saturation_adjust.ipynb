```python
import brushcue
from PIL import Image
import io

input_image_1 = brushcue.composition_monet_women_with_parasol() # insert your own image here
float_constant_2 = brushcue.float_constant(1.0)
saturation_scale_3 = brushcue.float_passthrough(float_constant_2)
saturation_adjust_4 = brushcue.composition_saturation_adjust(input_image_1, saturation_scale_3)

ctx = brushcue.Context()
result = saturation_adjust_4.execute(ctx)
composition = result.as_composition()
data_bytes = composition.to_image_bytes(ctx)
img = Image.open(io.BytesIO(data_bytes))
img.thumbnail((400, 400)) # remove this line for full resolution
img
```
