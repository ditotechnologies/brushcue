```python
import io
from PIL import Image

import brushcue as bc


@bc.brushcue_fn
def apply(input_image, strength):
    graded = input_image.color_transformer_shader(
        "let l = clamp(input.x, 0.0, 1.0);\nlet chroma = length(input.yz);\nlet blue = smoothstep(0.0, 0.055, -input.z) * smoothstep(0.008, 0.045, chroma);\nlet teal = smoothstep(0.0, 0.045, -input.y) * smoothstep(0.005, 0.045, chroma);\nlet ink = clamp(blue + teal * 0.7, 0.0, 1.0);\nlet levels = 9.0;\nlet x = l * levels;\nlet cut = (floor(x) + smoothstep(0.20, 0.80, fract(x))) / levels;\nlet tone = mix(l, cut, 0.65) - 0.045 * ink * (1.0 - l);\nlet ab = mix(input.yz * 0.72, vec2f(-0.023, -0.10) * (0.5 + 0.5 * chroma / 0.15), ink * 0.78);\nlet paper = smoothstep(0.75, 1.0, l);\nreturn vec4f(mix(input.xyz, vec3f(tone, ab + vec2f(0.002, 0.014) * paper), amount), input.w);",
        "",
        bc.ColorRepresentation.oklab_a(),
        bc.ColorRepresentation.oklab_a(),
        bc.Dictionary.create().add("amount", strength),
    )
    sampling = graded.color_transformer_shader(
        "return vec4f(input.rgb * input.a, input.a);",
        "",
        bc.ColorRepresentation.srgb(),
        bc.ColorRepresentation.srgb(),
        bc.Dictionary.create(),
    )
    bounds = input_image.bounds()
    reference_scale = (bounds.width().min(bounds.height()) / 1000.0).max(0.001)
    radius = (2.0 * reference_scale).max(0.5)
    return sampling.spacial_effect_shader(
        "let lo = image_bounds.xy + vec2f(0.5);\nlet hi = image_bounds.zw - vec2f(0.5);\nlet center_sample = sample(clamp(position, lo, hi));\nlet center = vec4f(center_sample.rgb / max(center_sample.a, 0.00001), center_sample.a);\nif (center.a <= 0.0) { return center; }\nvar total = vec3f(0.0);\nvar weight = 0.0;\nfor (var y = -1; y <= 1; y = y + 1) {\n  for (var x = -1; x <= 1; x = x + 1) {\n    let offset = vec2f(f32(x), f32(y));\n    let raw = sample(clamp(position + offset * radius, lo, hi));\n    let c = vec4f(raw.rgb / max(raw.a, 0.00001), raw.a);\n    let w = c.a * exp(-dot(c.rgb - center.rgb, c.rgb - center.rgb) * 90.0);\n    total += c.rgb * w; weight += w;\n  }\n}\nlet local = total / max(weight, 0.00001);\nreturn vec4f(mix(center.rgb, local, 0.55 * amount), center.a);",
        "",
        radius,
        bc.Dictionary.create()
        .add("amount", strength)
        .add("radius", radius)
        .add("image_bounds", bounds)
        .add("reference_scale", reference_scale),
        bc.ColorRepresentation.srgb(),
    ).crop(bounds)


strength = 1.0
input_image = bc.Composition.monet_women_with_parasol()
graph = apply(input_image, strength)

ctx = bc.Context()
composition = graph.execute(ctx)
data_bytes = composition.to_image_bytes(ctx)
img = Image.open(io.BytesIO(data_bytes))
img.thumbnail((400, 400))  # Remove this line for full resolution
img
```
